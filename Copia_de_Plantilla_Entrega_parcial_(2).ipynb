### 5) Análisis de complejidad **[15 pts]**
Incluye:
- Justificación
- Función T[n]
- Mejor y peor caso con gráfica


Justificación:
La red se modela sobre una lista de V vértices (estaciones) y una matriz de adyacencia de V \times V.  
Espacio: O(V^2) para almacenar la matriz de adyacencia y O(V) para los arreglos auxiliares (visitados, distancia, padre, cola, pila).  
Tiempo BFS: Para procesar un vértice se remueve de la cola en O(1) y se recorre su fila completa en la matriz de adyacencia de tamaño V. Por tanto, el recorrido BFS toma O(V^2) tiempo en el peor caso.
Tiempo DFS: De igual forma, cada vértice se visita una única vez y se inspeccionan sus V posibles vecinos en la matriz, tomando O(V^2) tiempo.
Función T(n) (donde n = V representa el número de vértices/estaciones):
Inicialización: T_{init}(n) = c_1 \cdot n^2 + c_2 \cdot n  
Recorrido BFS / Búsqueda de Ruta: T_{BFS}(n) = n \cdot (c_3 + c_4 \cdot n) = c_4 n^2 + c_3 n
Recorrido DFS: T_{DFS}(n) = n \cdot (c_5 + c_6 \cdot n) = c_6 n^2 + c_5 n
Función Total de Complejidad: T(n) = a \cdot n^2 + b \cdot n + c \implies \mathbf{\Theta(n^2)} (o \mathbf{\Theta(V^2)}).  
Mejor y Peor Caso:
Peor Caso (\mathcal{O}(V^2)): Ocurre cuando el grafo está completamente conectado (denso) o la búsqueda debe recorrer todos los nodos hasta el último nivel para encontrar el destino/completar el recorrido. Todas las celdas de la matriz son inspeccionadas.
Mejor Caso (\Omega(1) / \Omega(V)): Para la búsqueda de ruta, el mejor caso ocurre cuando el origen es igual al destino (O(1)) o cuando el destino es un vecino directo e inmediato en la primera iteración (O(V) para revisar la primera fila).

## 6) Código funcional **[15 pts]**

```python




def agregar_estacion(estaciones, nombre):
    if nombre not in estaciones:
        estaciones.append(nombre)

def obtener_indice(estaciones, nombre):
    for i in range(len(estaciones)):
        if estaciones[i] == nombre:
            return i
    return -1

def agregar_conexion(estaciones, matriz, origen, destino):
    agregar_estacion(estaciones, origen)
    agregar_estacion(estaciones, destino)

    pos_u = obtener_indice(estaciones, origen)
    pos_v = obtener_indice(estaciones, destino)

    matriz[pos_u][pos_v] = 1
    matriz[pos_v][pos_u] = 1



def mostrar_red(estaciones, matriz):
    print("\n--- REPRESENTACIÓN DE LA RED (LISTA DE ADYACENCIA) ---")
    tam = len(estaciones)
    for i in range(tam):
        vecinos = []
        for j in range(tam):
            if matriz[i][j] == 1:
                vecinos.append(estaciones[j])
        print(estaciones[i] + ": " + ", ".join(vecinos))



def bfs(estaciones, matriz, inicio):
    idx_start = obtener_indice(estaciones, inicio)
    if idx_start == -1:
        print("\nError: La estación '" + inicio + "' no existe en la red.")
        return

    n = len(estaciones)
    visitados = [False] * n
    distancia = [-1] * n

    cola = []

    visitados[idx_start] = True
    distancia[idx_start] = 0
    cola.append(idx_start)

    orden_visita = []

    while len(cola) > 0:
        actual = cola.pop(0)  # saca el primero
        orden_visita.append(estaciones[actual])

        for k in range(n):
            if matriz[actual][k] == 1 and not visitados[k]:
                visitados[k] = True
                distancia[k] = distancia[actual] + 1
                cola.append(k)

    print("\n--- RECORRIDO BFS DESDE '" + inicio + "' ---")
    print("Recorrido: " + " -> ".join(orden_visita))

    max_d = max(distancia)

    print("Niveles:")
    for niv in range(max_d + 1):
        est_nivel = []
        for i in range(n):
            if distancia[i] == niv:
                est_nivel.append(estaciones[i])
        print("  Nivel " + str(niv) + ": " + ", ".join(est_nivel))



def dfs(estaciones, matriz, inicio):
    idx_start = obtener_indice(estaciones, inicio)
    if idx_start == -1:
        print("\nError: La estación '" + inicio + "' no existe en la red.")
        return

    n = len(estaciones)
    visitados = [False] * n
    pila = [idx_start]
    orden_visita = []

    while len(pila) > 0:
        actual = pila.pop()  # saca el ultimo
        if not visitados[actual]:
            visitados[actual] = True
            orden_visita.append(estaciones[actual])

            for k in range(n - 1, -1, -1):
                if matriz[actual][k] == 1 and not visitados[k]:
                    pila.append(k)

    print("\n--- RECORRIDO DFS DESDE '" + inicio + "' ---")
    print("Recorrido: " + " -> ".join(orden_visita))



def buscar_ruta(estaciones, matriz, origen, destino):
    pos_origen = obtener_indice(estaciones, origen)
    pos_destino = obtener_indice(estaciones, destino)

    if pos_origen == -1 or pos_destino == -1:
        print("\nError: Una o ambas estaciones no existen en la red.")
        return

    n = len(estaciones)
    visitados = [False] * n
    padre = [-1] * n

    cola = [pos_origen]
    visitados[pos_origen] = True
    hallado = False

    while len(cola) > 0:
        actual = cola.pop(0)
        if actual == pos_destino:
            hallado = True
            break

        for k in range(n):
            if matriz[actual][k] == 1 and not visitados[k]:
                visitados[k] = True
                padre[k] = actual
                cola.append(k)

    if hallado:
        camino = []
        curr = pos_destino
        while curr != -1:
            camino.append(estaciones[curr])
            curr = padre[curr]

        camino.reverse()

        print("\n--- RUTA MÁS CORTA DE '" + origen + "' A '" + destino + "' ---")
        print("Ruta encontrada: " + " -> ".join(camino))
        print("Cantidad de conexiones: " + str(len(camino) - 1))
    else:
        print("\nNo existe una ruta entre '" + origen + "' y '" + destino + "'.")



def main():
    MAX_NODOS = 20
    estaciones = []
    matriz = [[0 for _ in range(MAX_NODOS)] for _ in range(MAX_NODOS)]

    conexiones_base = [
        ("Central", "Norte"),
        ("Central", "Museo"),
        ("Norte", "Parque"),
        ("Norte", "Universidad"),
        ("Museo", "Hospital"),
        ("Universidad", "Terminal"),
        ("Hospital", "Terminal")
    ]

    for u, v in conexiones_base:
        agregar_conexion(estaciones, matriz, u, v)

    # Salidas requeridas
    mostrar_red(estaciones, matriz)
    bfs(estaciones, matriz, "Central")
    dfs(estaciones, matriz, "Central")

    print("\n--- PUNTO 4: RECORRIDOS DESDE ESTACIÓN INICIAL LIBRE ---")
    libre = "Universidad"
    bfs(estaciones, matriz, libre)
    dfs(estaciones, matriz, libre)

    buscar_ruta(estaciones, matriz, "Central", "Terminal")

    print("\n--- PUNTO 6: AGREGAR NUEVA ESTACIÓN (Aeropuerto) ---")
    agregar_conexion(estaciones, matriz, "Aeropuerto", "Terminal")
    agregar_conexion(estaciones, matriz, "Aeropuerto", "Hospital")

    mostrar_red(estaciones, matriz)
    buscar_ruta(estaciones, matriz, "Central", "Aeropuerto")

main()
```
